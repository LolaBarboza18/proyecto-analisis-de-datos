```python
%pip install python-dotenv

import os
from dotenv import load_dotenv
import pandas as pd
from sqlalchemy import create_engine, text

load_dotenv()
DB_URI = os.getenv("SUPABASE_DB_URI")
engine = create_engine(DB_URI)

query = """
    SELECT title, date_published, genre, country, language, description, avg_vote
    FROM peliculas_raw 
    WHERE 
        description ILIKE '%mytholog%' OR 
        description ILIKE '%taoism%' OR
        description ILIKE '%martial arts%' OR
        description ILIKE '%dynasty%' OR
        description ILIKE '%monkey king%'
"""

with engine.connect() as connection:
    df_filtrado = pd.read_sql(text(query), con=connection)

df_limpio = df_filtrado.dropna(subset=['avg_vote']).copy()
df_limpio['genre'] = df_limpio['genre'].astype(str).str.lower()

df_limpio.to_sql('peliculas_procesadas', con=engine, if_exists='replace', index=False)

print(f"Se filtraron {len(df_limpio)} filas y se guardó la tabla 'peliculas_procesadas' en Supabase.")
```
